```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../data/pre_match_game_1.csv")
data
```

```python
fig, ax = plt.subplots(nrows=8, ncols=1, figsize=(10,20))

"""
later on this can just be a for loop
edits to robot code should be made where shooting can be shown as a boolean
we can separate through color on a graph when the shooter is running or not for better debugging
"""

ax[0].plot(data['time_seconds'], data['rf_power'])
ax[0].set_xlabel('Time (seconds)')
ax[0].set_ylabel('Right Front Wheel Power')

ax[1].plot(data['time_seconds'], data['lf_power'])
ax[1].set_xlabel('Time (seconds)')
ax[1].set_ylabel('Left Front Wheel Power')

ax[2].plot(data['time_seconds'], data['rb_power'])
ax[2].set_xlabel('Time (seconds)')
ax[2].set_ylabel('Right Back Wheel Power')

ax[3].plot(data['time_seconds'], data['lb_power'])
ax[3].set_xlabel('Time (seconds)')
ax[3].set_ylabel('Left Back Wheel Power')

# the next three graphs use Velocity instead of power
ax[4].plot(data['time_seconds'], data['shoot_power'])
ax[4].set_xlabel('Time (seconds)')
ax[4].set_ylabel('Shoot Motor Power')

ax[5].plot(data['time_seconds'], data['stopper_power'])
ax[5].set_xlabel('Time (seconds)')
ax[5].set_ylabel('Stopper Motor Power')

ax[6].plot(data['time_seconds'], data['it_power'])
ax[6].set_xlabel('Time (seconds)')
ax[6].set_ylabel('Intake Motor Power')

# battery voltage for possible regression model
ax[7].plot(data['time_seconds'], data['battery_voltage'])
ax[7].set_xlabel('Time (seconds)')
ax[7].set_ylabel('Battery Voltage (V)')

plt.tight_layout()
plt.show()
```
