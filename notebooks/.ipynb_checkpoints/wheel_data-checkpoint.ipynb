```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
wheel_df = pd.read_csv("../data/data_drivetrain_telemetry_20260925_154447.csv")
wheel_df
```

```python
fig, ax = plt.subplots(nrows=5, ncols=1, figsize=(10,20))

ax[0].plot(wheel_df['time_seconds'], wheel_df['rf_power'])
ax[0].set_xlabel('Time (seconds)')
ax[0].set_ylabel('Right Front Wheel Power')

ax[1].plot(wheel_df['time_seconds'], wheel_df['lf_power'])
ax[1].set_xlabel('Time (seconds)')
ax[1].set_ylabel('Left Front Wheel Power')

ax[2].plot(wheel_df['time_seconds'], wheel_df['rb_power'])
ax[2].set_xlabel('Time (seconds)')
ax[2].set_ylabel('Right Back Wheel Power')

ax[3].plot(wheel_df['time_seconds'], wheel_df['lb_power'])
ax[3].set_xlabel('Time (seconds)')
ax[3].set_ylabel('Left Back Wheel Power')

ax[4].plot(wheel_df['time_seconds'], wheel_df['battery_voltage'])
ax[4].set_xlabel('Time (seconds)')
ax[4].set_ylabel('Battery Voltage (V)')
```
